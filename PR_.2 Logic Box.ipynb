```python
while True:
    print("welcome to the pattern generator and number analyzer")
    print()
    while True:
        print("select and option")
        print("1. generate a pattern")
        print("2. analyze a range of patterns")
        print("3. exit")

        choice = int(input("enter your choice: "))

        if choice == 1:
            print("user select to generate pattern")
            numbers = int(input("enter the number of rows for the pattern: "))
            print("pattern")
            for i in range(1, numbers + 1):
                print('* ' * i)

        elif choice == 2:
            print("user select pattern analyzer")
            start_range = int(input("start range: "))
            end_range = int(input("end range: "))
            total = 0
            for i in range(start_range, end_range+1):
                if i % 2 == 0:
                    print("number", i, "is even")
                else:
                    print("number", i, "is odd")
                total += i
                i += 1
                
            print("total sum of the numbers are: ", total)
        
        elif choice == 3:
            print("goood bye, visit again ")

            break
        else:
            print("invalid choice try again")
    break
```
